### 3.3 Especificación del equivalente IaaS

*Se diseña, no se implementa. Máquinas y dimensionamiento, sistema operativo,
software a instalar, red, almacenamiento, y estimación del esfuerzo de puesta
en marcha y de operación.*


**Propuesta de infraestructura:**

| Componente | Propuesta de infraestructura |
|---|---|
| Servidor de procesamiento | Máquina virtual de 4 vCPU y 16 GB de RAM |
| Servidor de almacenamiento | Máquina virtual o almacenamiento de 500 GB |
| Sistema operativo | Linux Ubuntu Server |
| Software | Spark, Python y PostgreSQL |
| Red | Red privada con acceso controlado |
| Almacenamiento | Discos SSD para los datos |
| Seguridad | Usuarios, contraseñas, permisos y reglas para el acceso |

**ESFUERZO QUE TENDREMOS**

Cuando configuremos nos pedira que  instalemos y configuremos el sistema
operativo despues lo que es el Spark que es la  herramientas de procesamiento,  la red, el
almacenamiento y los permisos.

El esfuerzo  sería mayor que en una plataforma que fuese  administrada como
Databricks, porque el equipo tendría que trabajar  del mantenimiento, tambien de las 
actualizaciones, de la seguridad, la  capacidad de almacenamiento y disponibilidad
de las máquinas.
### 3.4 Comparación IaaS / PaaS / SaaS

| Criterio | IaaS | PaaS | SaaS |
|---|---|---|---|
| Control |Podemos administrar maquinas, sistemas operativos y el software el control es ALTO |Podemos administrar principalmente la aplicacion y los datos, el control es MEDIO |El proveedor puede administrar practicamente toda las soluciones, el control es BAJO |
| Tiempo hasta el primer resultado |Se re quiere que configuremos la infraestructura, el tiempo es MAYOR |Gran parte de la infraestructura ya esta activa o preparada asi que el tiempo es MENOR |El servidor esta listo para que lo utilicemos por eso el tiempo es BAJO |
| Esfuerzo operativo |Requiere lo que es mantenimiento, actualizaciones y la seguridad, el esfuerzo es ALTO |El proveedor administra la parte de la infraestructura, por eso el esfuerzo es MEDIO |El proveedor se va encargar de la mayoria de las operaciones y procesos por eso el esfuerzo es BAJO |
| Costo |Depende de las maquinas y de los recursos que utilicemos |Depende de los servicios y recursos que utilicemos  |Se paga suscripcion por el uso del servicio |
| Escalabilidad |Requiere configurar y ampliar los recursos |Permite que escalemos utilizando servios disponibles de la plataforma |El proveedor administra lo que es la escalabilidad de el servicio |
| Gobierno |Mayor control sobre nuestra infraestructura |control medio de los datos, los usuarios, y de la aplicacion |El proveedor controla parte de la infraestructura de los servicios |

**CONCLUSION Y MEJOR ELECION:**
IaaS nos da  mayor control, pero también requiere mayor esfuerzo
operativo. SaaS baja  más la administración, aunque ofrece menos
control sobre la solución. PaaS se encuentra en un intermedio entre lo que es
control y facilidad de administración.
